#Lists
Ordered collection of items; allows duplicates.


```python
# Define a list of data science jobs
job_skills = ['sql','tableau','excel']
job_skills
```

####Indexing
What if we want to get a specific item in a list? We'd use list indexing.
Lists are indexed, which means each item has a numerical position, so you can access it by referring to the index number.
Note: The first item has index 0.

```python
# Get a specific item in the list
job_skills[0]
```

####Change Value
To change the value of a specific item, refer to its index number.


```python
# Change the value of an item
job_skills[1] ='bigquery'

job_skills
```

##Methods

###`Append()`
`Append()` can be used to add an item at the end of the list.

```python
## Add a job to the list
job_skills.append('looker')

job_skills
```

###`Length()`
`Len()` returns number of items in the list.


```python
len(job_skills)
```

###`Insert`
To insert a list item at a specified index.

```python
# Insert an item into the list
job_skills.insert(2,'python')

job_skills
```

###`Remove()`
To remove a specified item from the list.

```python
# Remove an item from the list
job_skills.remove('looker')

job_skills
```

###Join Lists

1.   Concatenate using the `+` operator

```python
# Concatenate
skills1 = ['SQL', 'Tableau']
skills2 = ['Excel']

skills3 = skills1 + skills2
skills3
```

2. Appending all items from first list to second using `append()`

```python
# Append
skills4 = ['Python', 'Power BI']

# Append all items from skills3 to skills4 list
for x in skills3:
    skills4.append(x)

skills4
```

3. Use `extend()` method to add elements from one list to another list.

```python
# Extend
skills5 = ['Statistics', 'Machine Learning']

skills5.extend(skills4)

skills5
```

4. `Join()`

`.join()` is used for concatentating a list of strings together with a specified separator.

```python
skills = ['Python', 'SQL', 'Excel']
# Use concatenate print formatting
print('I have these skills: ' +  ', '.join(skills))
```

###Slicing Lists
Syntax: `list[start:stop:step]`

`start` has default value of 0 and `stop` has default value of the last value in the list, `step` has default value of 1.

```python
skills = ['Python', 'SQL', 'Excel']

# Extract the first two items
first_two = skills[0:2]
first_two
```

```python
full_list = skills[:]
full_list
```

```python
also_first_two = skills[:2]
also_first_two
```

```python
last_two = skills[1:]
last_two
```

```python
last_one = skills[-3:]
last_one
```

```python
skills = ['Python', 'SQL', 'Excel', 'R', 'Java']

# Extract every second item starting from the first
every_second = skills[0::2]
every_second
```

### Unpack List
To unpack a list (unpacking is when you assign each value in a list to a variable in a single statement) you can assign the list elements to variables directly.

```python
# Unpacking the list
skill1, skill2, skill3, skill4 = job_skills

# Printing the unpacked variables
print(skill1)
print(skill2)
print(skill3)
print(skill4)
```

####Extend Unpack List
To extend unpack a list, is when you assign a subset of elements to a variable as a list.

```python
# Unpacking the sql skills together and then unpacking the rest of the skills
*sql_skills, skill3, skill4 = job_skills

print(sql_skills)  # List of the SQL skills
print(skill3)
print(skill4)
```

#Sets

*   Unique Items
*   For storing multiple items in a single variable
*   It is:
    * Ordered
    * Unchangeable
    * Unindexed
*   Set *items* are unchangeable but items can be added or removed





```python
# Define a set of data science skills
job_skills = {'tableau', 'sql', 'python', 'statistics'}

job_skills
```

###Unordered and No Index
We can't use index to access items

```python
job_skills[1]
```

###Add()

```python
# Adding a skill
job_skills.add('looker')

job_skills
```

Adding an item that already exists in the set doesn't create duplicates.

```python
# Adding a skill that already exists doesn't duplicate
job_skills.add('sql')

job_skills
```

###Remove()

```python
# Removing a skill
job_skills.remove('statistics')

job_skills
```

###Set()
Sets are great for removing duplicates.

```python
# make a list of skills but some repeated values
skill_list = ['python', 'sql', 'statistics', 'tableau', 'python', 'sql', 'statistics', 'tableau']

set(skill_list)
```

###List()
This set can be converted back to a list. But it wil be created without the duplicate values.

```python
skill_list = list(set(skill_list))

skill_list
```

#Tuples

*   Used to store multiple items in a single variable
*   It is:
    *   Ordered
    *   Unchangeable
*   Written with (and)
*   Speeds up operation because it is immutable




```python
# Define a tuple of data science skills
job_skills = ('python', 'sql', 'statistics', 'tableau')

job_skills
```

##Get Item
To get an item from tuple, use indexing.

```python
# Accessing an element
job_skills[0]
```

##Slicing
To only get part of a the tuple, use slicing.

```python
# Define a set of data science skills
job_skills = ('python', 'sql', 'statistics', 'tableau')

job_skills[:2]
```

##Add Items
Since, the tuples are immutable, they don't have a built-in `append()` function.
But there are two methods:


**Method 1:** Turn tuple into a list
*   Convert the tuple into list
*   Use `append()`
*   Convert list back to tuple



```python
# Turn tuple into a list
job_skills_list = list(job_skills)

# Add skill to the list
job_skills_list.append('excel')

# Convert list to a tuple
job_skills_tuple = tuple(job_skills_list)

job_skills_tuple
```

**Method 2:** Add tuple to a tuple
*   Create a new tuple with the items
*   Add it to the existing tuple

```python
job_skills_new_tuple = ('r', 'postgresql',)
job_skills += job_skills_new_tuple

job_skills
```

###`Join()`
For tuples, we are combining multiple strings from a tuple.

```python
skills = ('Python', 'SQL', 'Excel')
# f-string formatting
print(f'I have these skills: {", ".join(skills)}')
```

##Remove Items

###Remove()
Convert the tuple into a list, remove the item using `remove()` method, and convert it back to a tuple.

```python
# Turn tuple into a list
job_skills_remove = list(job_skills)

# Remove skill from the list
job_skills_remove.remove('tableau')

# Convert list to a tuple
job_skills_tuple = tuple(job_skills_remove)

job_skills_tuple
```

###Del
Use the del keyword to delete the tuple completely.

```python
del job_skills
job_skills
```
