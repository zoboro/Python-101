# Python Terms

*   Object
*   Variable
*   Function
*   Class
*   Method
*   Atribute

## Object: The Data Records
An object is a specific instance of a class. It contains data (attributes) and behaviors (methods).

```python
name = "Alice"
```

In this example, `"Alice"` is a string object.

```python
class Dog:
    pass

my_dog = Dog()
```

`my_dog` is an object of the `Dog` class.

## Variable: The Reference to Objects
A variable is a name that refers to (points to) an object in memory.

```python
age = 25
```

Here, `age` is a variable that refers to the integer object `25`.

##Function: The Manipulators of Objects
A function is a reusable block of code that performs a task.

```python
def greet(name):
    return f"Hello, {name}!"
message = greet("Alice")
print(message)
```

Functions help avoid repeating code.

##Class: The Templates of Objects
A class is a blueprint for creating objects. It defines what data and behaviors objects will have.

```python
class Dog:
    def bark(self):
        print("Woof!")
```

The class describes what a object can do.

##Method: The Functions of an Object
A method is a function that belongs to a class and operates on its objects.

```python
class Dog:
    def bark(self):
        print("Woof!")
```

`bark()` is a method

```python
my_dog = Dog()
my_dog.bark()
```

Methods are functions associated with objects.

##Attribute: The Variables of an Object
An attribute is a piece of data stored inside an object.

```python
class Dog:
    def __init__(self, name):
        self.name = name
#Creating an object
my_dog = Dog("Buddy")
#Accessing the attribute
print(my_dog.name)
```

Here, `name` is an attribute of the `my_dog` object.

```python
class Dog:                    # Class
    def __init__(self, name):
        self.name = name      # Attribute

    def bark(self):           # Method
        print(f"{self.name} says Woof!")

my_dog = Dog("Buddy")         # Object
my_dog.bark()                 # Method call
```

| Term      | Example        | Description                   |
| --------- | -------------- | ----------------------------- |
| Variable  | `my_dog`       | Name that refers to an object |
| Object    | `Dog("Buddy")` | Instance of a class           |
| Class     | `Dog`          | Blueprint for objects         |
| Attribute | `name`         | Data stored in an object      |
| Function  | `greet()`      | Reusable block of code        |
| Method    | `bark()`       | Function belonging to a class |
