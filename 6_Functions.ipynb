#Functions


*   A **function** is a block of code that runs only when it is called.
*   You can pass **parameters** into a function.
*   Can be reused and make it more modular.




|Type of Function	|Example Function|
|-----------------|----------------|
|Built-In functions|	max()|
|User-defined functions|	def my_function(): pass|
|Lambda functions|	lambda x: x + 1|

Few popular types of functions are:

##User Defined Functions
Functions created by programmers using the `def` keyword.

**WARNING**

Do not name the function with the same name as the built-in function. In such case, the built-in function would be overridden.

```python
import builtins

def len(x):
    return 100

print(len([1, 2, 3]))           # 100
print(builtins.len([1, 2, 3]))  # 3
```

* Exit a function at a certain point.
* It can return multiple values using a tuple, list, or dictionary.

###Arguments
* Information can be passed into functions as arguments, inside the parantheses ().
* A function has to be called with the correct number of arguments. If your function has 2 arguments, you have to call the function with exactly 2 arguments.

####Parameters and Arguments

Parameters and arguments can be used for the same thing. Information that's passed into a function.

| Term          | Meaning                                   |
| ------------- | ----------------------------------------- |
| **Parameter** | A variable listed in a function definition.|
| **Argument**  | The actual value passed to the function when it is called.|


```python
def greet(name):  # 'name' is a parameter
    print(f"Hello, {name}!")

greet("Alice")    # "Alice" is an argument
```

###Keyword Arguments
* You can send arguments with key = value syntax
* Then the order of the arguments doesn't matter

```python
def introduce(name, age):
    print(f"Name: {name}, Age: {age}")

introduce(name="Alice", age=25)
```

####Uses of Keyword Arguments



```python
##Improves Readabililty

introduce("Alice", 25)

#vs#

introduce(name="Alice", age=25)
```

```python
##Order doesn't matter

introduce("Alice", 25)  # Correct

introduce(25, "Alice")  # Incorrect order

##with keyword arguments

introduce(age=25, name="Alice")
```

```python
##Useful with default parameters

def create_account(username, role="user", active=True):
    print(username, role, active)

create_account("john", active=False)
```

* `john` for username
* default `user` for role
* `False` for active (Overwrites True from the function)

###Arbitrary Arguments
Use `*args` when you don't know how many positional arguments will be passed.

```python
def total(*args):
    print(args)
    return sum(args)

print(total(10, 20, 30))
```

* `*args` collects all positional arguments into a tuple.
* args is just a convention; you can use another name:

```python
def total(*numbers):
    print(numbers)
    return  max(numbers)

total(1, 2, 3)
```

###Arbitrary Keyword Arguments
Use `**kwargs` when you don't know how many keyword arguments will be passed.

* `**kwargs` collects all keyword arguments into a dictionary.
* Keys are parameter names and values are the passed values.

```python
def display(**kwargs):
    print(kwargs)

display(name="Alice", age=25, city="Mumbai")
```

##Lambda Functions
Small, single-expression functions created with the `lambda` keyword.

*   They can have any number of arguments but only one expression.
*   Syntax `lambda arguments: expression`
*   Since its anonymous it doesn't require a separate name like regular functions are defined using def keyword (it's good for when you need a function temporarily



```python
##Regular Function

def square(x):
    return x * x

print(square(5))

##Lambda

square = lambda x: x * x

print(square(5))
```

```python
##Multiple arguments
add = lambda a, b: a + b

print(add(10, 20))
```

```python
##With sort()
students = [
    ("Alice", 85),
    ("Bob", 72),
    ("Charlie", 90)
]

students.sort(key=lambda student: student[1])

print(students)
```

```python
##With map()
numbers = [1, 2, 3, 4]

squares = list(map(lambda x: x * x, numbers))

print(squares)
```

```python
##With filter()
numbers = [1, 2, 3, 4, 5, 6]

evens = list(filter(lambda x: x % 2 == 0, numbers))

print(evens)
```

```python
##With reduce()
from functools import reduce

result = reduce(lambda a, b: a + b, [1, 2, 3, 4])

print(result)
```

###Create a Module

Save the code in the file with file extension `.py`




```python

%%writefile job_analyzer.py

# Call the function from the module
def calculate_salary(base_salary, bonus_rate=.1):
  """
  Calculate the total salary based on the base salary and bonus rate.

  Args:
    base_salary (float): The base salary.
    bonus_rate (float): The bonus rate. Default is .1.

  Returns:
    float: The total salary.
  """
  return base_salary * (1 + bonus_rate)

def calculate_bonus(total_salary, base_salary):
  """
  Calculate the bonus rate based on the total salary and base salary.

  Args:
    total_salary (float): The total salary.
    base_salary (float): The base salary.

  Returns:
    float: The bonus rate.
  """
  return (total_salary - base_salary) / base_salary
```
