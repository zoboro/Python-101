#Loops
* `while` loops
*  `for` loops

Used for automating repetitive tasks.

##`while` loop
* `while` loop executes a set of statements as long as the condition is true.
* Remember to increment `i`, or else the loop will continue forever.

```python
count = 1
while count <= 5:
    print(count)
    count += 1
```

```python
# Total number of job positions to check
total_positions = 3
position_index = 0

# Minimum years of experience required for each data science job position
position_experience_requirements = [1, 3, 2]
while position_index < total_positions:
    print('Position requires', position_experience_requirements[position_index], 'years of experience.')
    position_index += 1
```

```python
# User's years of experience
user_years_of_experience = 2

user_years_of_experience
```

```python
position_index = 0
while position_index < total_positions:
    required_years = position_experience_requirements[position_index]
    if user_years_of_experience >= required_years:
        print('Qualified')
    else:
        print('Not qualified')
    position_index += 1
```

##`for` loop


*   `for` loop iterates over a sequence (e.g. list, tuple, dictionary, set, or string).
*   Execute a set of statements, one for each item in a list, tuple, set, etc.



```python
for x in position_experience_requirements:
    print(f'Position requires {x} years of experience.')
```

###Enumerate


*   Used in loops
*   Used when you need an index of the iterable object for operations like:
    * updating the elements based on their position
    * comparing items at different positions
    * displaying the position along with them
*   Syntax `enumerate(iterable, start=0)`

Instead of iterating over the list with `x in list` to get the years of experience required for each position, we use `enumerate` to get both index and the years of experience from position experience requirements.


```python
# Enhanced example using enumerate to include the position index
for index, years in enumerate(position_experience_requirements, start=1):
    print(f'Position {index} requires {years} years of experience.')
```


###Pass
`pass` is used when a loop has nothing in it, to avoid errors.


```python
for x in position_experience_requirements:
    pass
```

###Nested Loops
These are loops inside another loop. Below we're going through two lists.


```python
roles = ['Data Scientist', 'Machine Learning Engineer']
skills = ['Python', 'SQL', 'Machine Learning']

for role in roles:
    print(f'For the role of {role}, you need experience in:')
    for skill in skills:
        print(f'  - {skill}')
```

```python
# For loop to check qualification for each job
for required_experience in position_experience_requirements:
    if user_years_of_experience >= required_experience:
        print('Qualified')
    else:
        print('Not qualified')
```

###Break
The break statement can stop the loop before it's looped through all of the items.

```python
# For loop to check qualification for each job
for required_experience in position_experience_requirements:
    if user_years_of_experience >= required_experience:
        break
    else:
        print('Not qualified')
```

##range() function


*   `range()` runs a `for` loop a specific number of times.
*   It returns a sequence of numbers, starting from 0 by default, and increments by 1 (default), and ends at the specified number.



```python
for x in range(4):
    print(x)
```

```python

for x in range(1,6,2):
    print(x)
```
