#Operators

##Types of Operators
Operators are used to perform operations on variables and values.

| Operator Type |Operators                           |
| ------------- | -----------------------------------|
| Arithmetic    | `+`, `-`, `*`, `/`, `%`, `//`, `**`|
| Comparison    | `==`, `!=`, `>`, `<`, `>=`, `<=`   |
| Assignment    | `=`, `+=`, `-=`, `*=`, `/=`        |
| Logical       | `and`, `or`, `not`                 |
| Identity      | `is`, `is not`                     |
| Membership    | `in`, `not in`                     |
| Bitwise       | `&`, `\|`,`^`, `~`, `<<`, `>>`     |


###Arithmetic Operators
Used with objects to perform math operations:

| Operator | Meaning             | Example         |
| -------- | ------------------- | --------------- |
| `+`      | Addition            | `5 + 2` → `7`   |
| `-`      | Subtraction         | `5 - 2` → `3`   |
| `*`      | Multiplication      | `5 * 2` → `10`  |
| `/`      | Division            | `5 / 2` → `2.5` |
| `//`     | Floor Division      | `5 // 2` → `2`  |
| `%`      | Modulus (remainder) | `5 % 2` → `1`   |
| `**`     | Exponent            | `5 ** 2` → `25` |


####Order of Operations
1.   **Parentheses**: Anything in parentheses is executed first
2.   **Exponents**: Operations involving exponents (including roots, indicated as fractional exponents)
3.   **Multiplication and Division**: Processed from left to right
4.   **Addition and Subtraction**: Also processed from left to right





####Some operations can be performed on non-numeric values

```python
"How are you? " * 10
```

```python
"How are " + "you?"
```

```python
#Strings don't have a subtract method
"How are you?" - "you?"
```

###Assignment Operators
Used to assign values to variables.

| Operator | Name                | Example |
|----------|---------------------|---------|
| =        | Assign              | x = y   |
| +=       | Add and assign      | x += y  |
| -=       | Subtract and assign | x -= y  |
| *=       | Multiply and assign | x *= y  |
| /=       | Divide and assign   | x /= y  |
| %=       | Modulus and assign  | x %= y  |
| **=      | Exponent and assign | x **= y |
| //=      | Floor divide and assign | x //= y  |

```python
#Swap Values
x=1
y=2

x, y = y, x
print("x is now: ", x)
print("y is now: ", y)
```

Add and Assign

```python
x = 1 # let's increment x by 1

x = x + 1 # not pythonic

x
```

```python
x = 1 # let's increment x by 1

x += 1 # pythonic

x
```

###Comparison Operators
Used to compare values. Result is `True` or `False`.



| Operator | Name                  | Example |
|----------|-----------------------|---------|
| ==       | Equal to              | x == y  |
| !=       | Not equal to          | x != y  |
| >        | Greater than          | x > y   |
| <        | Less than             | x < y   |
| >=       | Greater than or equal | x >= y  |
| <=       | Less than or equal    | x <= y  |


```python
salary_kelly = 50_000
salary_luke = 50_000

# Are Luke and Kelly's salaries the same?
salary_checker = (salary_kelly == salary_luke) # parantheses are optional

salary_checker
```

```python
type(salary_checker)
```

###Logical Operators
Used for combining conditional statements and evaluate the truth or falsity of the conditions.

```python
job_status_luke = not True

job_status_luke
```

```python
job_status_luke = False
job_status_kelly = True

# Do both Luke and Kelly have jobs?
job_status_luke and job_status_kelly
```

```python
# Do either Luke or Kelly have a job?
job_status_luke or job_status_kelly
```

###Identity Operators
Used for comparing object identities. Checks if two variables occupy same memory location i.e. are identical.


|Operator|Name|Example|
|--------|----|-------|
|is	     |Identity	|x is y|
|is not  |Not identity|	x is not y|


```python
salary_kelly = 100_000
salary_luke = 100_000

salary_kelly is salary_luke
```

```python
lukes_skills = ['Python', 'R', 'SQL']
kellys_skills = lukes_skills

lukes_skills is kellys_skills
```

###Membership Operators
Tests for membership in a sequence, such as strings, lists, or tuples.

|Operator|Name|Example|
|--------|----|-------|
|in      |Membership	|x in y|
|not in  |Not membership|	x not in y|

```python
'Data' in 'Data Nerd'
```

```python
'data' in 'Data Nerd'
```

```python
100 in [100, 200, 300]
```

```python
'Python' in ['Python', 'R', 'SQL']
```

###Bitwise Operators
Performs bit-by-bit operations on binary numbers.


|Operator |Name       |Example|
|---------|----       |-------|
|&	      |Bitwise AND|	x & y |
|\|	      |Bitwise OR	|x \| y |
|^	      |Bitwise XOR|x ^ y  |
|~	      |Bitwise NOT|~x     |
|<<	      |Left Shift	|x << y |
|>>	      |Right Shift|x >> y |


You could technically use & and | in place of and and or for Boolean comparison, BUT it's less readable.

You'll see the & and | operators pop up again when we cover Pandas

```python
True & False
```

```python
True | False
```

```python
x = 10  # binary 1010
y = 6   # binary 0110
result = x & y  # binary 0010 -> decimal 2

result
```

```python
x = 10  # binary 1010
y = 6   # binary 0110
result = x | y  # binary 1110 -> decimal 14

result
```
