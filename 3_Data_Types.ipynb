# Data Types
A data type tells Python what kind of value is being stored and what operations can be performed on it.

## Built-in Types

Python has the following data types built-in by default:
| Type     | Data Type    | Example               | Definition |
|----------|--------------|-----------------------|------------|
| Text     | `str`        | `"Data Nerd!"`        |Used to store textual data (characters and strings)|
| Numeric  | `int`        | `42`                  |Whole numbers without decimals|
|          | `float`      | `3.14159`             |Numbers with decimal points|
|          | `complex`    | `1 + 2j`              |Numbers with a real and imaginary part|
| Sequence | `list`       | `[1, 2, 3]`           |Ordered, mutable (can be changed)|
|          | `tuple`      | `(1, 2, 3)`           |Ordered, immutable (cannot be changed)|
|          | `range`      | `range(10)`           |Represents a sequence of numbers|
| Mapping  | `dict`       | `{"key": "value"}`    |Stores data as key-value pairs|
| Set      | `set`        | `{1, 2, 3}`           |Unordered collection of unique elements|
|          | `frozenset`  | `frozenset([1, 2, 3])`|Immutable version of a set|
| Boolean  | `bool`       | `True` or `False`     |Represents logical values|
| Binary   | `bytes`      | `b"Data"`             |Immutable sequence of bytes|
|          | `bytearray`  | `bytearray(5)`        |Mutable sequence of bytes|
|          | `memoryview` | `memoryview(b"Data")` |Provides direct access to binary data without copying|
| None     | `NoneType`   | `None`                |Represents the absence of a value|

##Check Data Types
Use type() to see the data type.

You can also check to see if a variable is the type you think it is. Using `isinstance()`. It has two arguments:
1. *Object* - what you want to check
2. *Type* - a type (what you want to use to see the object is)

For example, if I wanted to check if the `job_id` is a `float` type. If it is the type you put in, it will return `True` if not, it will return `False`.


```python
job_id = 102
company_name = 'Data Nerd, Inc.'
job_title = 'Data Scientist'
salary_rate = 170000.00
job_work_from_home = True
```

```python
isinstance(job_id, float)
```

```python
isinstance(company_name, str)
```

###Magic Methods (aka Dunder Methods)
Dunder methods (short for "double underscore methods") are special functions that override behavior for Python objects.



####Adding

```python
# don't write code like this,
# this is just to show the dunder method
str.__add__("Data", " Analyst")
```

```python
# still not Pythonic to write it this way
"Data".__add__(" Analyst")
```

Dunder methods are sometimes overridden by operators; such as in this case +

```python
# This is how you should use the __add__() method
"Data" + " Analyst"
```

####Length

```python
skill = "Python"

# once again, don't write code this way, this is for demonstrations only:
skill.__len__()
```

```python
# instead this is how to invoke the .__len__() method:
len(skill)
```

##String Formatting

In Python there's a few different options for formatting our output:

| Method          | Syntax                                        | Example (prints "Hello Luke")                                   |
|-----------------|------------------------------------------------|------------------------------------------------|
| Concatenation   | `'string' + variable`                          | `'Hello ' + name`                              |
| % Formatting    | `'string %s' % variable`                       | `'Hello %s' % name`                            |
| str.format()    | `'string {}'.format(variable)`                 | `'Hello {}'.format(name)`                      |
| f-strings       | `f'string {variable}'`                         | `f'Hello {name}'`                              |
| join()          | `'separator'.join([string, variable])`         | `', '.join(['Hello', name])`                   |

For our examples, we will be printing out a `role` and/or `skill`

```python
role = "Data Analyst"
skill = "Python"
```

###Types

####Concatenation `+`
Concatenation involves combining strings with the + operator.

```python
'Role: ' + role
```

####%-Formatting (Not Preffered)
%-formatting is an older method where `%` symbols serve as placeholders for variables.
| Specifier | Meaning   |
| --------- | --------- |
| `%s`      | String    |
| `%d`      | Integer   |
| `%f`      | Float     |
| `%c`      | Character |


```python
'Role: %s, Skill required: %s' % (role, skill)
```

####str.format()
The `str.format()` method uses curly braces `{}` as placeholders. It's more flexible than %-formatting and allows for reordering and reuse of arguments.

```python
'Role: {}, Skill required: {}'.format(role, skill)
```

```python
'Role: {role}, Skill required: {skill}'.format(
    role="Data Nerd", skill="SQL")
```

####f-Strings (Formatted String Literals - Preferred)
Use f-strings by prefixing a string with f and placing variables or expressions in {} braces directly within the string.
They are both concise and efficient.


```python
f'Role: {role}, Skill required: {skill}'
```

####Using `join()`
The `join()` method is useful for concatenating iterable items (e.g., strings, lists) into a single string, often separated by a specified character.

```python
years_expeirence = "0123456789"

', '.join(years_expeirence)
```

```python
skills = ['Python', 'SQL']
'Role: ' + role + ', Skills required: ' + ', '.join(skills)
```
