##Date and Time

```python
##DateTime
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])

##Extract Year
df['year'] = df['job_posted_date'].dt.year

##Extract Month
df['month'] = df['job_posted_date'].dt.month

##Extract Day
df['day'] = df['job_posted_date'].dt.day

##Extract Date
df['date'] = df['job_posted_date'].dt.date

df
```

##Sorting Values
`sort_values()` sorts a DataFrame or a specific column in ascending or descending order based on one or more columns.
Parameters:
* `by` - column name or list of column names to sort by
* `ascending` - boolean or list of booleans, default True, to sort by descending you would use False
* `inplace` - whether to modify the DataFrame in place or return a new one

```python
##Sort DataFrame by the job_posted_date in descending order

df.sort_values(by='job_posted_date', ascending=False, inplace=True)
df
```

##Adding Column

```python
##If the column job_title_short is equal to 'Data Analyst' then then return 1 if not, return 0. It does this by using astype(int).
df['Is Data Analyst'] = (df.job_title_short == 'Data Analyst').astype(int)
df['Is Data Analyst']
```

##Dropping Data
Use `drop()` if you want to drop (delete) either a column or row in your database.

The syntax is:
* Drop column: `df.drop('column_name, axis = 1)`
* Drop row: `df.drop(index_name, axis = 0)`

If you wanted to drop multiple rows you would have the syntax:
* Drop multiple columns: `df.drop(['column_name1', 'column_name2'], axis=1)`
* Drop multiple rows: `df.drop([index_name1, index_name2], axis=0)`


```python
df.drop(['year','month','day','date'], axis = 1, inplace=True)
df.info()
```

###Common Data Analysis Methods
* `df.sum()` - Sum of values
* `df.cumsum()` - Cummulative sum of values
* `df.min()/df.max()` - Minimum/maximum values
* `df.idxmin()/df.idxmax()` - Indexes of minimum/Maximum value
* `df.mean()` - Mean of values
* `df.median()` - Median of values
* `df.mode()` - Mode of the values
* `df.nunique()` — Number of distinct (non-NA) values per column
* `series.value_counts()` - Counts of unique values for a Series (aka a column)
  * Typically used for counting unique occurrences within a single column.
  * It's not applicable to an entire DataFrame without specifying a column.
* Depending on the method, you can apply it directly to a DataFrame or a Series.

##Aggregation

###Groupby()
* Use groupby to group your DataFrame by the unique values in a specific column.
* Lets you perform aggregation (e.g. mean, sum) on the grouped data.
* Aggregations that can be performed:
`mean()`, `median()`, `sum()`, `min()`, `max()`, `count()`, `size()`, `first()`, `last()`, `unique()`, etc.

```python
df.groupby('job_title_short')['salary_year_avg'].mean()
```

```python
df.groupby('job_title_short')['salary_year_avg'].median()
```

###Agg()
`agg()` lets you apply multiple functions at once.

```python
df.groupby('job_title_short')['salary_year_avg'].agg(['min', 'max'])
```
