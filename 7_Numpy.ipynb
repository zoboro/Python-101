#Numpy
NumPy (Numerical Python) is a Python library used for fast numerical computations and working with arrays.

Some of the use cases of Numpy are: array operations, linear algebra, statistical functions, etc.


```python
import numpy as np
```

##Array Operations




###Creating Arrays


```python
##Using a list
arr = np.array([1, 2, 3, 4])
```

```python
##Using zeroes
np.zeros(5)
```

```python
##Using ones
np.ones(5)
```

```python
##Range of numbers
np.arange(1, 10)
```

###Basic Operations


```python
##Array Shape
arr = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print(arr.shape)

##Indexing
##1-D Array
arr = np.array([10, 20, 30, 40])

print(arr[0])

##2-D Array
arr = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print(arr[1, 2])
```

####Array Operations

```python
# Example: An array representing the number of years of experience required for three different data science job listings.
years_of_experience = np.array([1, 2, 3, 4, 5])
```

```python
##Addition
years_of_experience_plus_one = years_of_experience + 1
years_of_experience_plus_one
```

```python
##Subtraction
years_of_experience_minus_one = years_of_experience - 1
years_of_experience_minus_one
```

```python
##Multiplication
years_of_experience_double = years_of_experience * 2
years_of_experience_double
```

```python
##Division
years_of_experience_half = years_of_experience / 2
years_of_experience_half
```

```python
##Slicing
# Example: Selecting the experience requirement for the second and third job listings.
second_and_third_jobs_experience = years_of_experience[1:3]
second_and_third_jobs_experience
```

```python
##Boolean Indexing
# Example: Selecting only those job listings that require more than 1 year of experience.
jobs_with_more_than_one_year_exp = years_of_experience[years_of_experience > 2]
jobs_with_more_than_one_year_exp
```

####Mathematical Operations

```python
arr = np.array([1, 2, 3, 4])

##Sum

print(arr.sum())

##Product

print(arr.prod())

##Cumulative Sum

print(arr.cumsum())

##Cumulative Product

print(arr.cumprod())
```

####Statistical Operations

```python
##Mean

print(arr.mean())

##Maximum

print(arr.max())

##Minimum

print(arr.min())

##Variance

print(arr.var())

##Standard Deviation

print(arr.std())
```
